## Part 3 — Implement Gradient Descent

### Q5 — Write a reusable Gradient Descent function
Complete the function below. Use the update rules taught in class:

- **θ₀(new) = θ₀(old) − α × mean(ŷ − y)**
- **θ₁(new) = θ₁(old) − α × mean((ŷ − y) × x)**

Calculate **both new parameters before replacing either old parameter**. Calculate the cost **after each update**. The function must work for any supplied learning rate, iteration count, and initial parameters.

```python
def gradient_descent(x, y, theta0, theta1, alpha, iterations):
    # TODO (5.1): Store the number of observations in x -> m
    m = len(x)
    # TODO (5.2): Create an empty list -> cost_history
    cost_history = []
    # TODO (5.3): Write a for loop that repeats exactly iterations times
    for i in range(iterations):
        # TODO (5.4): Inside the loop, calculate theta0 + theta1*x -> y_pred
        y_pred = theta0 + theta1 * x
        # TODO (5.5): Calculate the next intercept using the current theta0 and theta1 -> new_theta0
        new_theta0 = theta0 - alpha * np.mean(y_pred - y)
        # TODO (5.6): Calculate the next slope using the SAME current theta0 and theta1 -> new_theta1
        new_theta1 = theta1 - alpha * np.mean((y_pred - y) * x)
        # TODO (5.7): Replace theta0 and theta1 with new_theta0 and new_theta1
        theta0 = new_theta0
        theta1 = new_theta1
        # TODO (5.8): Recalculate theta0 + theta1*x using the updated parameters -> y_pred
        y_pred = theta0 + theta1 * x
        # TODO (5.9): Calculate Σ(y_pred − y)² / (2*m) -> cost
        cost = np.sum((y_pred - y) ** 2) / (2 * m)
        # TODO (5.10): Append cost to cost_history
        cost_history.append(cost)
    # TODO (5.11): Return theta0, theta1, and cost_history in that order
    return theta0, theta1, cost_history

# TODO (5.12): Call gradient_descent(x, y, 0.0, 0.0, 0.01, 2000)
#              Store the three returned values -> gd_theta0, gd_theta1, cost_history
# TODO (5.13): Print gd_theta0, gd_theta1, and the last value of cost_history
# === Write your answer below: ===
gd_theta0, gd_theta1, cost_history = gradient_descent(x, y, 0.0, 0.0, 0.01, 2000)
print("gd_theta0 =", gd_theta0)
print("gd_theta1 =", gd_theta1)
print("Final cost =", cost_history[-1])
```

## Part 4 — Experiment 1: Learning rate

### Q8 — Compare different learning rates
Call the function from Q5 separately for **α = 0.001, 0.01, and 0.05**. For every run, use **θ₀ = 0.0, θ₁ = 0.0, and 2000 iterations**. Plot all three cost histories on one graph.

```python
# TODO (8.1): Create a list -> alpha_values = [0.001, 0.01, 0.05]
# TODO (8.2): Create an empty dictionary -> alpha_histories
# TODO (8.3): For each alpha in alpha_values, call gradient_descent(x, y, 0.0, 0.0, alpha, 2000)
# TODO (8.4): Store the returned cost history in alpha_histories with alpha as its key

# TODO (8.5): Plot cost_history using plt.plot().
#             Use label=f"alpha={alpha}" to identify the current learning rate.
#             Repeat this inside the loop so that all three cost curves appear on one graph.
# TODO (8.6): Set title "Effect of Learning Rate on Cost"
# TODO (8.7): Set x label "Iteration", y label "Cost J", and add a legend
# TODO (8.8): Display the plot using plt.show()
# === Write your answer here: ===
alpha_values = [0.001, 0.01, 0.05]
alpha_histories = {}
for alpha in alpha_values:
    _, _, cost_history = gradient_descent(x, y, 0.0, 0.0, alpha, 2000)
    alpha_histories[alpha] = cost_history
    plt.plot(cost_history, label=f"alpha={alpha}")
plt.title("Effect of Learning Rate on Cost")
plt.xlabel("Iteration")
plt.ylabel("Cost J")
plt.legend()
plt.show()
```
